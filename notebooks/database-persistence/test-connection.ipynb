```python
# Você pode começar a executar comandos postgres sql
# Criando tabelas no SGBD PostgreSQL
%%sql

CREATE TABLE IF NOT EXISTS correntista (
    conta_nr integer NOT NULL,
    nome_ag character varying(100),
    nome_cor character varying(100),
    endereco_cor character varying(200),
    cidade_c character varying(100),
    CONSTRAINT correntista_pkey PRIMARY KEY (conta_nr)
);
```

```python
# Você pode começar a executar comandos postgres sql
# Criando tabelas no SGBD PostgreSQL
%%sql

CREATE TABLE IF NOT EXISTS deposito (
    nr_dep integer NOT NULL,
    conta_nr integer,
    nome_ag character varying(100),
    nome_cor character varying(100),
    saldo numeric(12,2),
    CONSTRAINT deposito_pkey PRIMARY KEY (nr_dep),
    CONSTRAINT deposito_conta_nr_fkey FOREIGN KEY (conta_nr)
        REFERENCES correntista (conta_nr)
);
```

```python
#Fazendo uma consulta na tabela Correntista no SGBD postgreSQL
%%sql
SELECT DISTINCT dep1.nome_cor
FROM Deposito dep1
WHERE NOT EXISTS (
    SELECT nome_ag
    FROM correntista
    WHERE NOT EXISTS (
        SELECT 1
        FROM Deposito dep2
        WHERE dep2.nome_cor = dep1.nome_cor
    )
);
```
